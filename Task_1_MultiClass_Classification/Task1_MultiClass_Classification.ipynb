**Goal:** Build end-to-end classification pipelines with Scikit-Learn, data normalization (StandardScaler), 80/20 train/test splitting, Logistic Regression and Decision Tree models, and benchmark classification accuracy, precision, recall, and F1-score with confusion matrices and decision boundary visualization.


```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score

iris = load_iris()
X_tr, X_te, y_tr, y_te = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42, stratify=iris.target)
scaler = StandardScaler()
X_tr_s = scaler.fit_transform(X_tr)
X_te_s = scaler.transform(X_te)

lr = LogisticRegression(max_iter=200, random_state=42).fit(X_tr_s, y_tr)
dt = DecisionTreeClassifier(max_depth=4, random_state=42).fit(X_tr, y_tr)

print(f"Logistic Regression Accuracy: {accuracy_score(y_te, lr.predict(X_te_s))*100:.2f}%")
print(f"Decision Tree Accuracy: {accuracy_score(y_te, dt.predict(X_te))*100:.2f}%")
```
